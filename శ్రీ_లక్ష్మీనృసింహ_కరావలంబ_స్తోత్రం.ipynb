```python
import cv2
import numpy as np
import os
from gtts import gTTS
from PIL import Image, ImageFont, ImageDraw
from moviepy.editor import VideoFileClip, AudioFileClip, concatenate_videoclips
import textwrap # Import the textwrap module
# --- Configuration ---
OUTPUT_FOLDER = "stotra_video_output"
os.makedirs(OUTPUT_FOLDER, exist_ok=True)
TEMP_FOLDER = os.path.join(OUTPUT_FOLDER, "temp")
os.makedirs(TEMP_FOLDER, exist_ok=True)
# Define media paths
telugu_font_path = "font/static/NotoSerifTelugu-Regular.ttf"
image_d1 = "narasimha.png"

# --- Story Content (Telugu Script) ---
story_parts = [

    #1
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "శ్రీమత్పయోనిధినికేతనచక్రపాణే",
            "భోగీంద్రభోగమణిరాజితపుణ్యమూర్తే |",
            "యోగీశ శాశ్వత శరణ్య భవాబ్ధిపోత",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 1 ||"
        ],
        "image": image_d1
    },
    #2
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "బ్రహ్మేంద్రరుద్రమరుదర్కకిరీటకోటి-",
            "-సంఘట్టితాంఘ్రికమలామలకాంతికాంత |",
            "లక్ష్మీలసత్కుచసరోరుహరాజహంస",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 2 ||"

        ],
        "image": image_d1
    },
    #3
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "సంసారదావదహనాకరభీకరోరు-",
            "-జ్వాలావలీభిరతిదగ్ధతనూరుహస్య |",
            "త్వత్పాదపద్మసరసీరుహమాగతస్య",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 3 ||"
        ],
        "image": image_d1
    },
    #4
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "సంసారజాలపతితస్య జగన్నివాస",
            "సర్వేంద్రియార్థబడిశాగ్రఝషోపమస్య |",
            "ప్రోత్కంపితప్రచురతాలుకమస్తకస్య",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 4 ||"#2
        ],
        "image": image_d1
    },
    #5
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "సంసారకూపమతిఘోరమగాధమూలం",
            "సంప్రాప్య దుఃఖశతసర్పసమాకులస్య |",
            "దీనస్య దేవ కృపయా పదమాగతస్య",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 5 ||"
        ],
        "image": image_d1
    },
    #6
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "సంసారభీకరకరీంద్రకరాభిఘాత-",
            "-నిష్పీడ్యమానవపుషః సకలార్తినాశ |",
            "ప్రాణప్రయాణభవభీతిసమాకులస్య",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 6 ||"#3
        ],
        "image": image_d1
    },
    #7
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "సంసారసర్పవిషదగ్ధమహోగ్రతీవ్ర-",
            "-దంష్ట్రాగ్రకోటిపరిదష్టవినష్టమూర్తేః |",
            "నాగారివాహన సుధాబ్ధినివాస శౌరే",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 7 ||"
        ],
        "image": image_d1
    },
    #8
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "సంసారవృక్షమఘబీజమనంతకర్మ-",
            "-శాఖాయుతం కరణపత్రమనంగపుష్పం |",
            "ఆరుహ్య దుఃఖఫలితం పతతో దయాలో [చకితం] ",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 8 ||"#4
        ],
        "image": image_d1
    },
    #9
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "సంసారసాగరవిశాలకరాలకాల-",
            "-నక్రగ్రహగ్రసితనిగ్రహవిగ్రహస్య |",
            "వ్యగ్రస్య రాగనిచయోర్మినిపీడితస్య",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 9 ||"
        ],
        "image": image_d1
    },
    #10
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "సంసారసాగరనిమజ్జనముహ్యమానం",
            "దీనం విలోకయ విభో కరుణానిధే మాం |",
            "ప్రహ్లాదఖేదపరిహారపరావతార",#5
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 10 ||"
        ],
        "image": image_d1
    },
    #11
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "సంసారఘోరగహనే చరతో మురారే",
            "మారోగ్రభీకరమృగప్రచురార్దితస్య |",
            "ఆర్తస్య మత్సరనిదాఘసుదుఃఖితస్య ",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 11 ||"
       ],
        "image": image_d1
    },
    #12
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "బద్ధ్వా గలే యమభటా బహుతర్జయంతః",
            "కర్షంతి యత్ర భవపాశశతైర్యుతం మాం |",
            "ఏకాకినం పరవశం చకితం దయాళో",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 12 ||"
        #6
        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "లక్ష్మీపతే కమలనాభ సురేశ విష్ణో",
            "యజ్ఞేశ యజ్ఞ మధుసూదన విశ్వరూప |",
            "బ్రహ్మణ్య కేశవ జనార్దన వాసుదేవ",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 13 ||"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "ఏకేన చక్రమపరేణ కరేణ శంఖ-",
            "-మన్యేన సింధుతనయామవలంబ్య తిష్ఠన్ |",
            "వామేతరేణ వరదాభయపద్మచిహ్నం",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 14 ||"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "అంధస్య మే హృతవివేకమహాధనస్య",
            "చోరైర్మహాబలిభిరింద్రియనామధేయైః |",
            "మోహాంధకారకుహరే వినిపాతితస్య",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 15 ||"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "ప్రహ్లాదనారదపరాశరపుండరీక-",
            "-వ్యాసాదిభాగవతపుంగవహృన్నివాస |",
            "భక్తానురక్తపరిపాలనపారిజాత",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 16 ||"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "లక్ష్మీనృసింహచరణాబ్జమధువ్రతేన",
            "స్తోత్రం కృతం శుభకరం భువి శంకరేణ |",
            "యే తత్పఠంతి మనుజా హరిభక్తియుక్తా-",
            "-స్తే యాంతి తత్పదసరోజమఖండరూపం || 17 ||"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "ఇతి శ్రీమత్పరమహంసపరివ్రాజకాచార్య శ్రీమచ్ఛంకరాచార్య విరచితం శ్రీలక్ష్మీనృసింహ కరావలంబ స్తోత్రం |"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "అధికశ్లోకాః:సంసారయోగ సకలేప్సితనిత్యకర్మ",
            "సంప్రాప్యదుఃఖ సకలేంద్రియమృత్యునాశ |",
            "సంకల్ప సింధుతనయాకుచ కుంకుమాంక",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 18 ||"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "ఆద్యంతశూన్యమజమవ్యయమప్రమేయం",
            "స్థితి, లయలకు మూలకారణమైన పరబ్రహ్మ స్వరూపుడిగా స్తుతిస్తుంది."
            "ఆదిత్యరుద్రనిగమాదినుతప్రభావం |",
            "అంభోధిజాస్య మధులోలుప మత్తభృంగ లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 19 ||"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "వారాహ రామ నరసింహ రమాదికాంతా",
            "క్రీడావిలోల విధిశూలి సురప్రవంద్య |",
            "హంసాత్మకం పరమహంస విహారలీలం",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 20 ||"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "మాతా నృసింహశ్చ పితా నృసింహః" ,
            "భ్రాతా నృసింహశ్చ సఖా నృసింహః |",
            "విద్యా నృసింహో ద్రవిణం నృసింహః",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 21 ||"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "ప్రహ్లాద మానస సరోజ విహారభృంగ",
            "గంగాతరంగ ధవళాంగ రమాస్థితాంక |",
            "శృంగార సుందర కిరీట లసద్వరాంగ",
            "లక్ష్మీనృసింహ మమ దేహి కరావలంబం || 22 ||"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "శ్రీశంకరార్య రచితం సతతం మనుష్యః",
            "స్తోత్రం పఠేదిహతు సత్వగుణప్రసన్నం |",
            "సద్యోవిముక్త కలుషో మునివర్య గణ్యో",
            "లక్ష్మీ పదముపైతి సనిర్మలాత్మా || 23 ||"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "యన్మాయయోర్జితః వపుః ప్రచుర ప్రవాహ",
            "మగ్నార్థ మత్రనివహోరు కరావలంబం |",
            "లక్ష్మీనృసింహ చరణాబ్జ మధువ్రతేన",
            "స్తోత్రం కృతం శుభకరం భువి శంకరేణ || 24 ||"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "శ్రీమన్నృసింహ విభవే గరుడధ్వజాయ",
            "తాపత్రయోపశమనాయ భవౌషధాయ |",
            "తృష్ణాది వృశ్చిక జలాగ్ని భుజంగ రోగ",
             "క్లేశాపహాయ హరయే గురవే నమస్తే || 25 ||"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "ఇతి శ్రీలక్ష్మీనృసింహ కరావలంబ స్తోత్రం |"

        ],
        "image": image_d1
    },
   #13
    {
        "title": "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం",
        "text": [
            "ఈ స్తోత్రంలో భక్తుడు లక్ష్మీ నరసింహుని తనకు శరణు ఇవ్వమని వినయంగా ప్రార్థిస్తాడు.",
            "నరసింహుడు భయంకర దుఃఖాలను, కష్టాలను తొలగించే మహాశక్తి స్వరూపుడు.",
            "ఆయన కరుణ దృష్టి పాపాలను నశింపజేసి భక్తునికి ధైర్యం ప్రసాదిస్తుంది.",
            "ఈ స్తోత్రం పఠిస్తే శత్రుభయం పోయి శాంతి, భక్తి, మోక్షాభిలాష పెరుగుతుంది."

        ],
        "image": image_d1
    }



]
    # --- Video Generation Functions ---
width, height = 1920, 1080
fps = 24

def create_frame(title, subtitle, image_path):
    """Creates a single video frame with text and image."""
    img = Image.new('RGB', (width, height), color='black')
    draw = ImageDraw.Draw(img)
#font
    try:
        font_size_title = 70
        font_size_subtitle = 35
        title_font = ImageFont.truetype(telugu_font_path, font_size_title)
        subtitle_font = ImageFont.truetype(telugu_font_path, font_size_subtitle)
    except IOError:
        print("Warning: Font file not found. Using default font.")
        title_font = ImageFont.load_default()
        subtitle_font = ImageFont.load_default()

    if os.path.exists(image_path):
        bg_image = Image.open(image_path).resize((int(width * 0.7), int(height * 0.7)), Image.LANCZOS)
        img.paste(bg_image, ((width - bg_image.width) // 2, (height - bg_image.height) // 2))

    if title:
        draw.text((width // 2, 50), title, font=title_font, fill='gold', anchor="mt")

    y_text = height - 200
    # Wrap text and draw each line
    max_width = width - 200 # Adjust this value as needed for padding
    wrapped_lines = []
    for line in subtitle:
        wrapped_lines.extend(textwrap.wrap(line, width=int(max_width / (font_size_subtitle * 0.6)))) # Adjust width based on font size and cast to integer

    for line in wrapped_lines:
        draw.text((width // 2, y_text), line, font=subtitle_font, fill='white', anchor="mt")
        y_text += 60 # Adjust line spacing as needed

    return cv2.cvtColor(np.array(img), cv2.COLOR_RGB2BGR)

def create_scene_video(scene_data, scene_index):
    """Creates a video clip for a single scene with synchronized audio."""
    voiceover_text = " ".join(scene_data["text"])
    audio_path = os.path.join(TEMP_FOLDER, f"scene_{scene_index}.mp3")

    if "Aparajita devia namaha" in voiceover_text:
        tts = gTTS(text=voiceover_text, lang='hi')
    else:
        tts = gTTS(text=voiceover_text, lang='te')

    tts.save(audio_path)

    audio_clip = AudioFileClip(audio_path)
    audio_duration = audio_clip.duration

    video_path = os.path.join(TEMP_FOLDER, f"scene_{scene_index}శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం.mp4")
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(video_path, fourcc, fps, (width, height))

    num_frames = int(audio_duration * fps)

    for _ in range(num_frames):
        frame = create_frame(scene_data["title"], scene_data["text"], scene_data["image"])
        out.write(frame)

    out.release()
    audio_clip.close()

    print(f"Created video for Scene {scene_index} with duration {audio_duration:.2f} seconds.")

    return video_path, audio_path

# --- Main Program Execution ---
if __name__ == "__main__":
    print("Starting video creation process...")

    video_clips = []

    for i, scene in enumerate(story_parts):
        try:
            video_path, audio_path = create_scene_video(scene, i)
            video_clip = VideoFileClip(video_path).set_audio(AudioFileClip(audio_path))
            video_clips.append(video_clip)

        except Exception as e:
            print(f"An error occurred while processing scene {i}: {e}")
            continue

    if video_clips:
        print("\nCombining all scenes into a final video...")
        final_video = concatenate_videoclips(video_clips)

        final_output_path = os.path.join(OUTPUT_FOLDER, "శ్రీ లక్ష్మీనృసింహ కరావలంబ స్తోత్రం.mp4")

        final_video.write_videofile(final_output_path, codec="libx264", audio_codec="aac")

        print(f"\n✅ Video created successfully at: {final_output_path}")

    else:
        print("\n❌ No video clips were created. Exiting.")

    import shutil
    shutil.rmtree(TEMP_FOLDER)
    print("Temporary files have been cleaned up.")

    try:
        from google.colab import files
        files.download(final_output_path)
    except ImportError:
        print("\nNote: The 'google.colab' library is not available. You can find your video file in the 'stotra_video_output' folder on the left-hand side file explorer.")
```
