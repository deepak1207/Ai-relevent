```python
import cv2
import numpy as np
import os
from gtts import gTTS
from PIL import Image, ImageFont, ImageDraw
from moviepy.editor import VideoFileClip, AudioFileClip, concatenate_videoclips
import textwrap # Import the textwrap module
# --- Configuration ---
OUTPUT_FOLDER = "stotra_video_output"
os.makedirs(OUTPUT_FOLDER, exist_ok=True)
TEMP_FOLDER = os.path.join(OUTPUT_FOLDER, "temp")
os.makedirs(TEMP_FOLDER, exist_ok=True)
# Define media paths
telugu_font_path = "font/static/NotoSerifTelugu-Regular.ttf"
image_d1 = "pocham.png"
image_d2 = "title.jpg"
# --- Story Content (Telugu Script) ---

    #1
story_parts = [
    {
        "title": "The Vedansh Devotional channel",
        "text": [
            "Welcome to My channel viewers ",

        ],
        "image": image_d2
    },

    {
        "title": "పోచమ్మ దేవాలయం",
        "text": [
            "పోచమ్మ తల్లి తెలంగాణ మరియు ఆంధ్రప్రదేశ్ గ్రామాల్లో విస్తృతంగా పూజించే",
            "గ్రామ దేవత (Gramadevata). ఆమెను గ్రామాన్ని కాపాడే శక్తివంతమైన అమ్మవారిగా భావిస్తారు."
        ],
        "image": image_d1
    },

    {
        "title": "పోచమ్మ దేవాలయం",
        "text": [
            "పోచమ్మ తల్లి మహిమ:-",
            "పోచమ్మ తల్లి ఆదిశక్తి అవతారంగా భావిస్తారు. భక్తులు ఆమెను ప్రధానంగా ఈ కారణాల కోసం పూజిస్తారు:",
            "వ్యాధుల నుండి రక్షణ (ప్రత్యేకంగా జ్వరాలు, చర్మ వ్యాధులు)",
            "ఆరోగ్యం మరియు ఆయురారోగ్యం"
        ],
        "image": image_d1
        },
         {
        "title": "పోచమ్మ దేవాలయం",
        "text": [
            "పోచమ్మ తల్లి మహిమ:-",
            "పోచమ్మ తల్లి ఆదిశక్తి అవతారంగా భావిస్తారు. భక్తులు ఆమెను ప్రధానంగా ఈ కారణాల కోసం పూజిస్తారు:",
            "గ్రామానికి శాంతి, సమృద్ధి",
            "దుష్ట శక్తుల నుండి కాపాడటం"
        ],
        "image": image_d1
        },
    {
        "title": "పోచమ్మ దేవాలయం",
        "text": [
            "దేవాలయ లక్షణాలు:-",
            "పోచమ్మ గుడులు సాధారణంగా గ్రామం బయట లేదా ప్రవేశ ద్వారం వద్ద ఉంటాయి",
            "ఆలయం సాదాసీదాగా ఉండి, శక్తి స్వరూపాన్ని ప్రతిబింబిస్తుంది",
            "కొన్ని చోట్ల రాతి విగ్రహం లేదా ప్రతీక రూపంలో పూజిస్తారు"
        ],
        "image": image_d1
        },
    {
        "title": "పోచమ్మ దేవాలయం",
        "text": [
            "పండుగలు & ఆచారాలు:-",
            "బోనాలు పండుగ (తెలంగాణ) పోచమ్మ తల్లికి ముఖ్యమైనది",
            "భక్తులు బోనం సమర్పించి అమ్మవారిని ప్రసన్నం చేసుకుంటారు"
        ],
        "image": image_d1
    },
    {
        "title": "పోచమ్మ దేవాలయం",
        "text": [
            "పండుగలు & ఆచారాలు:-",
            "ముఖ్య ఆచారాలు:1.పసుపు, కుంకుమ, నిమ్మ ఆకులు సమర్పించడం.",
            "2.జాతరలు, పల్లె పాటలు, నృత్యాలు.3.కొన్ని ప్రాంతాల్లో సంప్రదాయ బలులు."
        ],
        "image": image_d1
    },
    #9
    {
        "title": "పోచమ్మ దేవాలయం",
        "text": [
            "నమ్మకాలు:-",
            "పోచమ్మ తల్లి గ్రామాన్ని వ్యాధులు, అపశకునాల నుండి కాపాడుతుందని నమ్మకం",
            "భక్తులు మొక్కులు పెట్టి, నెరవేరిన తర్వాత నైవేద్యాలు సమర్పిస్తారు",
            "ఆమె పూజలో భక్తి ప్రధానమైనది, ఆచారాలు సాధారణంగా ఉంటాయి"
        ],
        "image": image_d1
    },
    {
        "title": "పోచమ్మ దేవాలయం",
        "text": [
            "ప్రసిద్ధ పోచమ్మ గుడులు:-",
            "హైదరాబాద్, సికింద్రాబాద్ పరిసర ప్రాంతాల్లో అనేక పోచమ్మ ఆలయాలు",
            "తెలంగాణలో ప్రతి గ్రామంలో దాదాపు ఒక పోచమ్మ గుడి ఉంటుంది"

        ],
        "image": image_d1
    }

]
    # --- Video Generation Functions ---
width, height = 1920, 1080
fps = 24

def create_frame(title, subtitle, image_path):
    """Creates a single video frame with text and image."""
    img = Image.new('RGB', (width, height), color='black')
    draw = ImageDraw.Draw(img)
#font
    try:
        font_size_title = 70
        font_size_subtitle = 30
        title_font = ImageFont.truetype(telugu_font_path, font_size_title)
        subtitle_font = ImageFont.truetype(telugu_font_path, font_size_subtitle)
    except IOError:
        print("Warning: Font file not found. Using default font.")
        title_font = ImageFont.load_default()
        subtitle_font = ImageFont.load_default()

    if os.path.exists(image_path):
        bg_image = Image.open(image_path).resize((int(width * 0.7), int(height * 0.7)), Image.LANCZOS)
        img.paste(bg_image, ((width - bg_image.width) // 2, (height - bg_image.height) // 2))

    if title:
        draw.text((width // 2, 50), title, font=title_font, fill='gold', anchor="mt")

    y_text = height - 200
    # Wrap text and draw each line
    max_width = width - 200 # Adjust this value as needed for padding
    wrapped_lines = []
    for line in subtitle:
        wrapped_lines.extend(textwrap.wrap(line, width=int(max_width / (font_size_subtitle * 0.6)))) # Adjust width based on font size and cast to integer
#change
    for line in wrapped_lines:#Deep Blue with Shadow
        draw.text((width // 2, y_text), line, font=subtitle_font, fill='green', anchor="mt")
        y_text += 60 # Adjust line spacing as needed

    return cv2.cvtColor(np.array(img), cv2.COLOR_RGB2BGR)

def create_scene_video(scene_data, scene_index):
    """Creates a video clip for a single scene with synchronized audio."""
    voiceover_text = " ".join(scene_data["text"])
    audio_path = os.path.join(TEMP_FOLDER, f"scene_{scene_index}.mp3")

    if "Aparajita devia namaha" in voiceover_text:
        tts = gTTS(text=voiceover_text, lang='hi')
    else:
        tts = gTTS(text=voiceover_text, lang='te')

    tts.save(audio_path)

    audio_clip = AudioFileClip(audio_path)
    audio_duration = audio_clip.duration

    video_path = os.path.join(TEMP_FOLDER, f"scene_{scene_index}పోచమ్మ దేవాలయం.mp4")
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(video_path, fourcc, fps, (width, height))

    num_frames = int(audio_duration * fps)

    for _ in range(num_frames):
        frame = create_frame(scene_data["title"], scene_data["text"], scene_data["image"])
        out.write(frame)

    out.release()
    audio_clip.close()

    print(f"Created video for Scene {scene_index} with duration {audio_duration:.2f} seconds.")

    return video_path, audio_path

# --- Main Program Execution ---
if __name__ == "__main__":
    print("Starting video creation process...")

    video_clips = []

    for i, scene in enumerate(story_parts):
        try:
            video_path, audio_path = create_scene_video(scene, i)
            video_clip = VideoFileClip(video_path).set_audio(AudioFileClip(audio_path))
            video_clips.append(video_clip)

        except Exception as e:
            print(f"An error occurred while processing scene {i}: {e}")
            continue

    if video_clips:
        print("\nCombining all scenes into a final video...")
        final_video = concatenate_videoclips(video_clips)

        final_output_path = os.path.join(OUTPUT_FOLDER, "పోచమ్మ దేవాలయం.mp4")

        final_video.write_videofile(final_output_path, codec="libx264", audio_codec="aac")

        print(f"\n✅ Video created successfully at: {final_output_path}")

    else:
        print("\n❌ No video clips were created. Exiting.")

    import shutil
    shutil.rmtree(TEMP_FOLDER)
    print("Temporary files have been cleaned up.")

    try:
        from google.colab import files
        files.download(final_output_path)
    except ImportError:
        print("\nNote: The 'google.colab' library is not available. You can find your video file in the 'stotra_video_output' folder on the left-hand side file explorer.")
```

```python
import cv2
import numpy as np
import os
from gtts import gTTS
from PIL import Image, ImageFont, ImageDraw
from moviepy.editor import VideoFileClip, AudioFileClip, concatenate_videoclips
import textwrap # Import the textwrap module
# --- Configuration ---
OUTPUT_FOLDER = "stotra_video_output"
os.makedirs(OUTPUT_FOLDER, exist_ok=True)
TEMP_FOLDER = os.path.join(OUTPUT_FOLDER, "temp")
os.makedirs(TEMP_FOLDER, exist_ok=True)
# Define media paths
telugu_font_path = "font/static/NotoSerifTelugu-Regular.ttf"
image_d1 = "pocham.png"
image_d2 = "title.jpg"
image_d3 = "gramasth1.png"
image_d4 = "gramasth2.png"
image_d5 = "sthani.png"
image_d6 = "mahila bakth.png"
image_d7 = "pochamma.png"
image_d8 = "british sai1.png"
image_d9 = "british adi1.png"
image_d10 = "gramasthulu.png"
image_d11 = "gram-pry.png"
# --- Story Content (Telugu Script) ---

    #1
story_parts = [
    {
        "title": "The Vedansh Devotional channel",
        "text": [
            "Welcome to My channel viewers ",

        ],
        "image": image_d2
    },

    {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "కథ ప్రారంభం:-పాత హైదరాబాద్ – లష్కర్ ప్రాంతం, వ్యాధులు వ్యాపిస్తున్న సమయం",
            "గ్రామస్తుడు 1:అయ్యో! ఈ వ్యాధి రోజురోజుకీ పెరుగుతోంది… మన గ్రామం కాపాడేది ఎవరు?"
        ],
        "image": image_d3
    },

    {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "కథ ప్రారంభం:-పాత హైదరాబాద్ – లష్కర్ ప్రాంతం, వ్యాధులు వ్యాపిస్తున్న సమయం",
            "మహిళా భక్తురాలు:మన పోచమ్మ తల్లిని ప్రార్థిద్దాం… ఆమె మనల్ని తప్పకుండా కాపాడుతుంది!"
        ],
        "image": image_d6
        },
         {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "పోచమ్మ తల్లికి ప్రార్థన:-",
            "గ్రామస్తులు (ఒకస్వరంగా):అమ్మా పోచమ్మా! మా గ్రామాన్ని ఈ మహమ్మారి నుండి కాపాడమ్మా!"
        ],
        "image": image_d11
        },
    {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "(దివ్య కాంతి మధ్య పోచమ్మ తల్లి ప్రత్యక్షమవుతుంది)",
            "పోచమ్మ తల్లి:నా పిల్లలారా… భయపడకండి. మీ భక్తి నన్ను సంతోషపరిచింది.",
"మీరు నన్ను విశ్వాసంతో పూజిస్తే, ఈ వ్యాధి తగ్గిపోతుంది."
        ],
        "image": image_d7
        },
    {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "బోనం ఆచారం ప్రారంభం",
            "మహిళా భక్తురాలు:అమ్మా! మేమెలా పూజించాలి?"
        ],
        "image": image_d6
    },
    {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "బోనం ఆచారం ప్రారంభం",
            "పోచమ్మ తల్లి:నాకు అన్నం, పాలు, బెల్లంతో బోనం తయారు చేసి సమర్పించండి.",
            "పసుపు, కుంకుమతో నన్ను అలంకరించండి."
        ],
        "image": image_d7
    },
    #9
    {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "బోనాలు పండుగ ప్రారంభం",
            "గ్రామస్తుడు 2:ఇప్పటి నుంచి ప్రతి సంవత్సరం బోనాలు పండుగ జరుపుదాం!"
        ],
        "image": image_d4
    },
    {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "బోనాలు పండుగ ప్రారంభం",
            "మహిళా భక్తురాలు (ఆనందంగా):అమ్మా పోచమ్మ కృపతో మన గ్రామం కాపాడబడింది!"
        ],
        "image": image_d6
    },
    {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "గోల్కొండ / లష్కర్ కథ (బ్రిటిష్ కాలం)",
            "బ్రిటిష్ సైనికుడు:ఈ ప్రాంతంలో వ్యాధి విస్తరిస్తోంది… మేము ఏం చేయాలి?"
        ],
        "image": image_d8
    },
    {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "గోల్కొండ / లష్కర్ కథ (బ్రిటిష్ కాలం)",
            "స్థానికుడు:మా పోచమ్మ తల్లి ఉన్నది… ఆమెను ప్రార్థిస్తే రక్షణ ఉంటుంది."
        ],
        "image": image_d5
    },
    {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "గోల్కొండ / లష్కర్ కథ (బ్రిటిష్ కాలం)",
            "బ్రిటిష్ అధికారి:సరే… మీ దేవతను ప్రార్థిద్దాం."
        ],
        "image": image_d9
    },
    {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "అద్భుతం (కొద్ది రోజుల తర్వాత వ్యాధి తగ్గిపోతుంది)",
            "బ్రిటిష్ అధికారి:ఇది నిజంగా ఒక అద్భుతం!మీ పోచమ్మ దేవి చాలా శక్తివంతమైనది!"
        ],
        "image": image_d9
    },
    {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "ముగింపు",
            "గ్రామస్తులు:జై పోచమ్మ తల్లి!ప్రతి సంవత్సరం బోనాలు పండుగగా జరుపుకుంటూ అమ్మవారిని స్మరించుకుందాం!"
        ],
        "image": image_d10
    },
    {
        "title": "బోనాలు పోచమ్మ తల్లి కథ",
        "text": [
            "సారాంశం",
            "పోచమ్మ తల్లి భక్తుల విశ్వాసంతో ప్రసన్నమై,",
            "వ్యాధుల నుండి రక్షణ ఇచ్చి,బోనాలు పండుగకు ఆరంభం ఇచ్చిన గ్రామదేవతగా నిలిచింది."
        ],
        "image": image_d1
    }
]
    # --- Video Generation Functions ---
width, height = 1920, 1080
fps = 24

def create_frame(title, subtitle, image_path):
    """Creates a single video frame with text and image."""
    img = Image.new('RGB', (width, height), color='black')
    draw = ImageDraw.Draw(img)
#font
    try:
        font_size_title = 70
        font_size_subtitle = 30
        title_font = ImageFont.truetype(telugu_font_path, font_size_title)
        subtitle_font = ImageFont.truetype(telugu_font_path, font_size_subtitle)
    except IOError:
        print("Warning: Font file not found. Using default font.")
        title_font = ImageFont.load_default()
        subtitle_font = ImageFont.load_default()

    if os.path.exists(image_path):
        bg_image = Image.open(image_path).resize((int(width * 0.7), int(height * 0.7)), Image.LANCZOS)
        img.paste(bg_image, ((width - bg_image.width) // 2, (height - bg_image.height) // 2))

    if title:
        draw.text((width // 2, 50), title, font=title_font, fill='gold', anchor="mt")

    y_text = height - 200
    # Wrap text and draw each line
    max_width = width - 200 # Adjust this value as needed for padding
    wrapped_lines = []
    for line in subtitle:
        wrapped_lines.extend(textwrap.wrap(line, width=int(max_width / (font_size_subtitle * 0.6)))) # Adjust width based on font size and cast to integer
#change
    for line in wrapped_lines:#Deep Blue with Shadow
        draw.text((width // 2, y_text), line, font=subtitle_font, fill='white', anchor="mt")
        y_text += 60 # Adjust line spacing as needed

    return cv2.cvtColor(np.array(img), cv2.COLOR_RGB2BGR)

def create_scene_video(scene_data, scene_index):
    """Creates a video clip for a single scene with synchronized audio."""
    voiceover_text = " ".join(scene_data["text"])
    audio_path = os.path.join(TEMP_FOLDER, f"scene_{scene_index}.mp3")

    if "Aparajita devia namaha" in voiceover_text:
        tts = gTTS(text=voiceover_text, lang='hi')
    else:
        tts = gTTS(text=voiceover_text, lang='te')

    tts.save(audio_path)

    audio_clip = AudioFileClip(audio_path)
    audio_duration = audio_clip.duration

    video_path = os.path.join(TEMP_FOLDER, f"scene_{scene_index}బోనాలు పోచమ్మ తల్లి కథ.mp4")
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(video_path, fourcc, fps, (width, height))

    num_frames = int(audio_duration * fps)

    for _ in range(num_frames):
        frame = create_frame(scene_data["title"], scene_data["text"], scene_data["image"])
        out.write(frame)

    out.release()
    audio_clip.close()

    print(f"Created video for Scene {scene_index} with duration {audio_duration:.2f} seconds.")

    return video_path, audio_path

# --- Main Program Execution ---
if __name__ == "__main__":
    print("Starting video creation process...")

    video_clips = []

    for i, scene in enumerate(story_parts):
        try:
            video_path, audio_path = create_scene_video(scene, i)
            video_clip = VideoFileClip(video_path).set_audio(AudioFileClip(audio_path))
            video_clips.append(video_clip)

        except Exception as e:
            print(f"An error occurred while processing scene {i}: {e}")
            continue

    if video_clips:
        print("\nCombining all scenes into a final video...")
        final_video = concatenate_videoclips(video_clips)

        final_output_path = os.path.join(OUTPUT_FOLDER, "బోనాలు పోచమ్మ తల్లి కథ.mp4")

        final_video.write_videofile(final_output_path, codec="libx264", audio_codec="aac")

        print(f"\n✅ Video created successfully at: {final_output_path}")

    else:
        print("\n❌ No video clips were created. Exiting.")

    import shutil
    shutil.rmtree(TEMP_FOLDER)
    print("Temporary files have been cleaned up.")

    try:
        from google.colab import files
        files.download(final_output_path)
    except ImportError:
        print("\nNote: The 'google.colab' library is not available. You can find your video file in the 'stotra_video_output' folder on the left-hand side file explorer.")
```

```python
import cv2
import numpy as np
import os
from gtts import gTTS
from PIL import Image, ImageFont, ImageDraw
from moviepy.editor import VideoFileClip, AudioFileClip, concatenate_videoclips
import textwrap # Import the textwrap module
# --- Configuration ---
OUTPUT_FOLDER = "stotra_video_output"
os.makedirs(OUTPUT_FOLDER, exist_ok=True)
TEMP_FOLDER = os.path.join(OUTPUT_FOLDER, "temp")
os.makedirs(TEMP_FOLDER, exist_ok=True)
# Define media paths
telugu_font_path = "font/static/NotoSerifTelugu-Regular.ttf"
image_d1 = "nookalamma.png"
image_d2 = "title.jpg"

# --- Story Content (Telugu Script) ---

    #1
story_parts = [
    {
        "title": "The Vedansh Devotional channel",
        "text": [
            "Welcome to My channel viewers ",

        ],
        "image": image_d2
    },

    {
        "title": "నూకాలమ్మ తల్లి – విశాఖపట్నం",
        "text": [
            "నూకాలమ్మ తల్లి ఎవరు?",
            "శ్రీ నూకాలమ్మ తల్లి విశాఖపట్నం ప్రాంతంలోని ప్రముఖ గ్రామదేవత (Gramadevata)",
            "ఆమెను శక్తి స్వరూపిణిగా భావించి భక్తులు గొప్ప భక్తితో పూజిస్తారు.",
            "“నూక” అంటే బియ్యం / ఆహారం, అందుకే నూకాలమ్మను అన్నపూర్ణ దేవి అవతారంగా కూడా భావిస్తారు."
        ],
        "image": image_d1
    },

    {
        "title": "నూకాలమ్మ తల్లి – విశాఖపట్నం",
        "text": [
            "నూకాలమ్మ ఆలయం – స్థానం:",
            "విశాఖపట్నం (Vizag) లోని పలు ప్రాంతాల్లో నూకాలమ్మ గుడులు ఉన్నాయి",
            "ముఖ్యంగా:1.అనకాపల్లి నూకాలమ్మ జాతర చాలా ప్రసిద్ధి",
            "2.స్థానికంగా ప్రతి ప్రాంతంలో చిన్న పెద్ద ఆలయాలు ఉన్నాయి"
        ],
        "image": image_d1
        },
         {
        "title": "నూకాలమ్మ తల్లి – విశాఖపట్నం",
        "text": [
            "నూకాలమ్మ జాతర (Anakapalli Jatara):",
            "ఇది ఆంధ్రప్రదేశ్‌లో అత్యంత పెద్ద జాతరల్లో ఒకటి.",
            "ప్రత్యేకతలు:ఉగాది తర్వాత జరిగే మహా ఉత్సవం,లక్షలాది మంది భక్తులు పాల్గొంటారు",
            "అమ్మవారికి:1.బోనం (అన్నం నైవేద్యం),2.పసుపు, కుంకుమ,3.చీరలు సమర్పిస్తారు"
        ],
        "image": image_d1
        },
    {
        "title": "నూకాలమ్మ తల్లి – విశాఖపట్నం",
        "text": [
            "భక్తుల నమ్మకాలు:",
            "నూకాలమ్మ తల్లి:ఆరోగ్యం, ఐశ్వర్యం ప్రసాదిస్తుంది,కుటుంబానికి అన్నం కొరత లేకుండా కాపాడుతుంది,",
            "వ్యాధులను దూరం చేస్తుంది",
            "భక్తులు మొక్కులు పెట్టి, నెరవేరిన తర్వాత జాతరలో తీర్చుకుంటారు"
        ],
        "image": image_d1
        },
    {
        "title": "నూకాలమ్మ తల్లి – విశాఖపట్నం",
        "text": [
            "ఆలయ విశేషాలు:",
            "ఆలయం సాధారణంగా పల్లెటూరి సంప్రదాయంలో ఉంటుంది",
            "అమ్మవారి విగ్రహం ఉగ్ర రూపంలో ఉంటుంది",
            "పూజలు సాదాసీదాగా కానీ భక్తి ప్రధానంగా జరుగుతాయి"
        ],
        "image": image_d1
    },
    {
        "title": "నూకాలమ్మ తల్లి – విశాఖపట్నం",
        "text": [
            "పురాణ/స్థానిక కథ (సంక్షిప్తంగా):",
            "ఒకప్పుడు గ్రామంలో కరువు, వ్యాధులు వచ్చాయి.",
            "అప్పుడు నూకాలమ్మ తల్లి ప్రత్యక్షమై,"
        ],
        "image": image_d1
    },
    #9
    {
        "title": "నూకాలమ్మ తల్లి – విశాఖపట్నం",
        "text": [
            "పురాణ/స్థానిక కథ (సంక్షిప్తంగా):",
            "“నన్ను పూజిస్తే మీకు అన్నం, ఆరోగ్యం లభిస్తుంది” అని ఆశీర్వదించిందని చెబుతారు.",
            "అప్పటి నుంచి భక్తులు ఆమెను విశ్వాసంతో పూజిస్తున్నారు."
        ],
        "image": image_d1
    },
    {
        "title": "నూకాలమ్మ తల్లి – విశాఖపట్నం",
        "text": [
            "సారాంశం:",
            "నూకాలమ్మ తల్లి:అన్నపూర్ణ స్వరూపిణి,గ్రామ రక్షక దేవత",
            "విశాఖపట్నం ప్రాంతంలో అత్యంత శక్తివంతమైన అమ్మవారిలో ఒకరు"
        ],
        "image": image_d1
    }
]
    # --- Video Generation Functions ---
width, height = 1920, 1080
fps = 24

def create_frame(title, subtitle, image_path):
    """Creates a single video frame with text and image."""
    img = Image.new('RGB', (width, height), color='black')
    draw = ImageDraw.Draw(img)
#font
    try:
        font_size_title = 70
        font_size_subtitle = 30
        title_font = ImageFont.truetype(telugu_font_path, font_size_title)
        subtitle_font = ImageFont.truetype(telugu_font_path, font_size_subtitle)
    except IOError:
        print("Warning: Font file not found. Using default font.")
        title_font = ImageFont.load_default()
        subtitle_font = ImageFont.load_default()

    if os.path.exists(image_path):
        bg_image = Image.open(image_path).resize((int(width * 0.7), int(height * 0.7)), Image.LANCZOS)
        img.paste(bg_image, ((width - bg_image.width) // 2, (height - bg_image.height) // 2))

    if title:
        draw.text((width // 2, 50), title, font=title_font, fill='gold', anchor="mt")

    y_text = height - 200
    # Wrap text and draw each line
    max_width = width - 200 # Adjust this value as needed for padding
    wrapped_lines = []
    for line in subtitle:
        wrapped_lines.extend(textwrap.wrap(line, width=int(max_width / (font_size_subtitle * 0.6)))) # Adjust width based on font size and cast to integer
#change
    for line in wrapped_lines:#Deep Blue with Shadow
        draw.text((width // 2, y_text), line, font=subtitle_font, fill='green', anchor="mt")
        y_text += 60 # Adjust line spacing as needed

    return cv2.cvtColor(np.array(img), cv2.COLOR_RGB2BGR)

def create_scene_video(scene_data, scene_index):
    """Creates a video clip for a single scene with synchronized audio."""
    voiceover_text = " ".join(scene_data["text"])
    audio_path = os.path.join(TEMP_FOLDER, f"scene_{scene_index}.mp3")

    if "Aparajita devia namaha" in voiceover_text:
        tts = gTTS(text=voiceover_text, lang='hi')
    else:
        tts = gTTS(text=voiceover_text, lang='te')

    tts.save(audio_path)

    audio_clip = AudioFileClip(audio_path)
    audio_duration = audio_clip.duration

    video_path = os.path.join(TEMP_FOLDER, f"scene_{scene_index}నూకాలమ్మ తల్లి – విశాఖపట్నం.mp4")
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(video_path, fourcc, fps, (width, height))

    num_frames = int(audio_duration * fps)

    for _ in range(num_frames):
        frame = create_frame(scene_data["title"], scene_data["text"], scene_data["image"])
        out.write(frame)

    out.release()
    audio_clip.close()

    print(f"Created video for Scene {scene_index} with duration {audio_duration:.2f} seconds.")

    return video_path, audio_path

# --- Main Program Execution ---
if __name__ == "__main__":
    print("Starting video creation process...")

    video_clips = []

    for i, scene in enumerate(story_parts):
        try:
            video_path, audio_path = create_scene_video(scene, i)
            video_clip = VideoFileClip(video_path).set_audio(AudioFileClip(audio_path))
            video_clips.append(video_clip)

        except Exception as e:
            print(f"An error occurred while processing scene {i}: {e}")
            continue

    if video_clips:
        print("\nCombining all scenes into a final video...")
        final_video = concatenate_videoclips(video_clips)

        final_output_path = os.path.join(OUTPUT_FOLDER, "నూకాలమ్మ తల్లి – విశాఖపట్నం.mp4")

        final_video.write_videofile(final_output_path, codec="libx264", audio_codec="aac")

        print(f"\n✅ Video created successfully at: {final_output_path}")

    else:
        print("\n❌ No video clips were created. Exiting.")

    import shutil
    shutil.rmtree(TEMP_FOLDER)
    print("Temporary files have been cleaned up.")

    try:
        from google.colab import files
        files.download(final_output_path)
    except ImportError:
        print("\nNote: The 'google.colab' library is not available. You can find your video file in the 'stotra_video_output' folder on the left-hand side file explorer.")
```
